```python
import pandas as pd

colaboradores = pd.read_excel("Catalogo_Colaboradores.xlsx")
df = pd.DataFrame(colaboradores)

def crearcolumnas():
    df["SalarioDiario"] = df["Sueldo Neto Total"] / 30
    df["CostoInterno"] = df["Sueldo Neto IMSS"] + df["Sueldo Complementos"]

def limpieza():
    df.drop(columns=["Notas"], inplace=True)

def estadisticasreferencia():
    print("Salario diario promedio:", df["SalarioDiario"].mean())
    print("Sueldo neto IMSS promedio:", df["Sueldo Neto IMSS"].mean())
    print("Complementos promedio:", df["Sueldo Complementos"].mean())
    print("Costo interno promedio:", df["CostoInterno"].mean())

def actualiza():
    df.to_excel("archivocambiado.xlsx", index=False)

crearcolumnas()
limpieza()
estadisticasreferencia()
actualiza()
```
